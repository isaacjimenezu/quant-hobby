```python
import pandas as pd
import numpy as np
import yfinance as yf

tickers = yf.download(["SPY", "IWM", "EFA", "EEM", "EWW", "IEF", "TLT", "LQD", "GLD", "VNQ"], start = "2006-01-01", end = "2026-11-01")
tickers
```

```python
precios = tickers["Close"]
precios
```

```python
print(precios.shape)
print(precios.head())
print(precios.tail())
```

```python
#Guardar precios como CSV en la carpeta de data
precios.to_csv('data/precios_etfs.csv', index=True)
precios_csv = pd.read_csv('data/precios_etfs.csv', index_col=0, parse_dates=True)
precios_csv
```

```python
print(precios_csv.shape)
print(precios_csv.index)
```

```python
precios_csv.isna().sum()
```

```python
((precios_csv-precios_csv.shift(1))==0).sum()
```

```python
diff_precios = precios_csv - precios_csv.shift(1)
diff_precios_positivo = abs(diff_precios)
diff_precios_positivo.median()
```

```python
rend_precios = precios_csv.pct_change()
print(rend_precios.std())
```

```python
relation_1 = rend_precios.std()/abs(rend_precios).median()
relation_1
```

```python
print(rend_precios["SPY"].min())
print(rend_precios["SPY"].idxmin())
print(rend_precios["SPY"].min()/rend_precios.std()["SPY"])
```

```python
precios_sigmas = abs(rend_precios)/rend_precios.std()
(precios_sigmas>4).sum()
```
