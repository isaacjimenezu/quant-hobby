```python
import pandas as pd
import numpy as np
import yfinance as yf

spy = yf.download("SPY", start="2016-01-01", end="2026-11-01")
```

```python
cierre = spy["Close"]

rend_simple = lambda x: x/x.shift(1)-1
rend_log = lambda x: np.log(x/x.shift(1))

np.isclose(np.log(cierre.iloc[-1]/cierre.iloc[0]),rend_log(cierre).sum())
```

```python
cierre=cierre["SPY"]
type(cierre)

```

```python
#Cuanto rindió SPY desde enero del 2016 hasta octubre del 2026
cierre_log = rend_log(cierre)
np.exp(cierre_log.sum())-1
```

```python
#Cuanto rindió en promedio por año
r_total = np.exp(cierre_log.sum())-1
periodo_len = ((cierre.index[-1]-cierre.index[0]).days)/((365*3+366)/4)
r_anual = np.power((1+r_total),1/periodo_len)-1
```

```python
print(r_total)
print(r_anual)
print(periodo_len)
```
