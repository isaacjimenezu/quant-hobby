```python
import pandas as pd
import numpy as np
print(pd.__version__, np.__version__)
```

```python
precios = pd.DataFrame(
    {"A": [100, 102, 99, 101], "B": [50, 49, 51, 52]},
    index=pd.date_range("2026-11-02", periods=4, freq="B"),
)
precios
```

```python
precios.shift(1)
```

```python
rend_simple = precios/precios.shift(1)-1
rend_simple
```

```python
rend_log = np.log(precios/precios.shift(1))
#rend_log
rend_log.sum()
```
